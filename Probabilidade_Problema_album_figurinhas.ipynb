# Problema do Álbum de Figurinhas

Um álbum de figurinhas requer $N$ figurinhas diferentes para ser completado. Presumindo que as figurinhas estão distribuídas aleatoriamente, em média, quantas precisamos comprar para completar o álbum?

## Resolução analítica

Notação:

*   $N$: Número de figurinhas do álbum;
*   $p(k)$: probabilidade de obtermos uma figurinha faltante, dado que  já possuímos $k$ figurinhas do álbum ($0 \le k \le N$);
*   $\eta(k)$: número médio de figurinhas que precisamos comprar para obtermos uma figurinha faltante, dado que já possuímos $k$ figurinhas do álbum ($0 \le k < N$).


Nas condições do problema, temos as seguintes expressões para as probabilidades
$$
p(k) = \frac{N-k}{N},\ 0 \le k \le N;
$$
Assim, as probabilidades complementarem são dadas por
$$
\bar{p}(k) = 1 - p(k) = \frac{k}{N},\ 0 \le k \le N.
$$

Agora, o número médio de compras para adquirirmos uma $(k+1)$-ésima figurinha é dado pela soma infinita:

$$
\begin{split}
\eta(k) &= 1\times p(k) + 2\times \bar{p}(k)p(k) + 3\times \bar{p}(k)^2p(k) + \dots\\
 &= \sum_{j = 1}^\infty p(k)j\bar{p}(k)^{j-1}\\
 &= p(k)\sum_{j = 1}^\infty j\left(\frac{k}{N}\right)^{j-1}\\
 &= p(k)\frac{1}{(1-k/N)^2}\\
 &= \frac{N}{N-k}\ \ \ \ (k= 0, \dots, N-1).
 \end{split}
$$

<font color="grey">(Curiosamente, o número médio de figurinhas que precisamos comprar para obter uma nova é igual ao inverso da probabilidade de obter uma nova figurinha com uma única compra!)</font>

Portanto, o número médio de figurinhas a comprar é dado pela soma do número médio que precisamos comprar para obter a primeira figurinha, depois a segunda figurinha, depois a terceira figurinha etc:
$$
\begin{split}
M &= \sum_{k=0}^{N-1}\eta(k)\\
    &= N \times \left( \frac{1}{1} + \frac{1}{2} + \dots + \frac{1}{N} \right).
\end{split}
$$

<font color="grey">(Outra curiosidade está na aparição da série harmônica na resposta do problema!)</font>

```python
'''
Implementação da fórmula analítica para o número médio de compras.
'''
def media_compras(N = int):
    soma = 0
    for k in range(1,N+1):
        soma += 1/k
    return N*soma
```

```python
# Exemplo

import numpy as np

N = 100
m = media_compras(N)
m_inteiro = np.ceil(m)
print(f"Media de compras para completar um álbum com {N} figurinhas: {m_inteiro}")
```

## Simulação compoutacional

```python
import random
import numpy as np

N = 100 # número de figurinhas diferentes do álbum (variável)

num_simulacoes = 100*N # número de experimentos

lista_compras = [0 for _ in range(num_simulacoes)] # lista para armazenar o número de figurinhas compradas em cada simulação

for _ in range(num_simulacoes):

    album = [] # figurinhas diferentes adquiridas
    n_compras = 0 # número inicial de figurinhas compradas
    n_album = 0 # número de figurinhas diferentes adquiridas

    while n_album != N:
        n_compras += 1
        k = random.randint(1, N)
        if k in album:
            pass
        else:
            album.append(k)
            n_album += 1

    lista_compras[_] = n_compras

m_frac = sum(lista_compras)/num_simulacoes

m_int = np.ceil(m_frac)

print(f"Media de compras para completar um álbum com {N} figurinhas: {m_int}")
```
